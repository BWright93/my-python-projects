Create a lottery machine using Python that produces 6 random numbers. <br>
Then compares the results to a player's ticket and checks if they have won. <br>
Also allows the user to compare multiple tickets to the same set of lottery numbers.

```python
import random
import time
import sys

def check_lotto_numbers():
    
    print(f'''
        ==========
         Welcome!
        ==========
        ''')
    
    while True:
        
        # Ask user/player for their ticket numbers
        # Remember to convert string input to integers
        print('Please enter your chosen lottery numbers, separated by spaces: ')
        player_input = input(" ")
        

        try:
            user_numbers = [int(num) for num in player_input.split()]
            #print(user_numbers)

            # Check that 6 numbers were given by the user
            if len(user_numbers) != 6:
                print(f'Invalid! You entered {len(user_numbers)} numbers. Please enter exactly 6 numbers!\n')
                continue

            # Check that given numbers are in the correct range (1 to 59)
            out_of_range = [num for num in user_numbers if num < 1 or num > 59]
            if out_of_range:
                print(f'Invalid! Please enter numbers in range 1 to 59 only!\n')
                continue

            # Check for any duplicate numbers given by the user
            if len(user_numbers) != len(set(user_numbers)):
                print(f'Invalid! Duplicate number detected, please try again!\n')
                continue

            # Get winning numbers using the 'random' module then sort them in ascending order
            # Use 'seed' to ensure the same winning numbers for each ticket we check
            random.seed(0)
            winning_numbers = random.sample(range(1,60), 6)
            winning_numbers.sort()
            
            print('\nChecking numbers...\n')
            time.sleep(10)

            #print('\n')

            # Find and return matching numbers
            matching_numbers = [num for num in user_numbers if num in winning_numbers]

            print("Tonight's winning numbers are: ", winning_numbers)
            print("The ticket numbers are: ", user_numbers)
            #print('\n')

            if matching_numbers:
                print(f'The matching numbers are: {matching_numbers}')
            else:
                print('No match. Better luck next time!')

        # Error if non-numerical input is given
        except ValueError:
            print('Invalid! Enter valid numbers only!')
            continue

        # Ask to check different numbers
        new_game = input('\nCheck different ticket? Y or N? ')
    
        if new_game[0].upper() == 'Y':
            continue
            
        else:
            print(f'''
        ==========================
          Thank you for playing!
        ==========================
            ''')
            break

check_lotto_numbers()
```
